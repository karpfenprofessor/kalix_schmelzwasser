# Weather and discharge exploration

Initial descriptive plots. These show associations, not forecast performance.


```python
import matplotlib.pyplot as plt
import seaborn as sns

from kalix.data import load_daily_data, project_root

ROOT = project_root()
daily = load_daily_data(ROOT)
measurements = ["temperature_c", "precipitation_mm", "discharge_m3_s"]

sns.set_theme(style="whitegrid")
```

```python
axes = daily[measurements].plot(
    subplots=True,
    figsize=(12, 8),
    sharex=True,
    title=["Temperature (°C)", "Precipitation (mm/day)", "Discharge (m³/s)"],
    legend=False,
)
plt.tight_layout()
plt.show()
```

```python
monthly = daily["discharge_m3_s"].groupby(daily.index.month).mean()
ax = monthly.plot(marker="o", figsize=(9, 4))
ax.set(xlabel="Month", ylabel="Mean discharge (m³/s)", title="Seasonal discharge pattern")
ax.set_xticks(range(1, 13))
plt.tight_layout()
plt.show()
```
