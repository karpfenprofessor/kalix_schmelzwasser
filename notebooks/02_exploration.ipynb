# Weather and discharge exploration

Run `.\.tools\uv.exe run python -m kalix.prepare` from the project root before opening this notebook.
These plots use the shared daily calendar and preserve gaps. They show associations, not forecast performance.


```python
import matplotlib.pyplot as plt
import seaborn as sns

from kalix.data import load_processed_data, project_root

ROOT = project_root()
daily = load_processed_data(ROOT)
measurements = [
    "nikkaluokta_temperature_c",
    "tarfala_temperature_c",
    "precipitation_mm",
    "discharge_m3_s",
]

sns.set_theme(style="whitegrid")
```

```python
axes = daily[measurements].plot(
    subplots=True,
    figsize=(12, 10),
    sharex=True,
    title=[
        "Nikkaluokta temperature (°C)",
        "Tarfala temperature (°C)",
        "Precipitation (mm/day)",
        "Discharge (m³/s)",
    ],
    legend=False,
)
plt.tight_layout()
plt.show()
```

```python
temperatures = daily[["nikkaluokta_temperature_c", "tarfala_temperature_c"]].rename(
    columns={"nikkaluokta_temperature_c": "Nikkaluokta", "tarfala_temperature_c": "Tarfala"}
)
ax = temperatures.plot(figsize=(12, 4), title="Daily temperature at both stations")
ax.set(ylabel="Temperature (°C)")
plt.tight_layout()
plt.show()
```

```python
monthly = daily["discharge_m3_s"].groupby(daily.index.month).mean()
ax = monthly.plot(marker="o", figsize=(9, 4))
ax.set(xlabel="Month", ylabel="Mean discharge (m³/s)", title="Seasonal discharge pattern")
ax.set_xticks(range(1, 13))
plt.tight_layout()
plt.show()
```
